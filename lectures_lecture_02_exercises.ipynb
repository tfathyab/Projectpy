```python
# lectures/lecture_02/exercises.ipynb

#(append / insert / remove / pop / sort) سلّة المشارى المشاركاب — A1
cart = []
cart.append("Latte")
cart.append("Croissant")
cart.append("Tea")
cart.remove("Croissant")
cart.insert(0, "Water")
cart.pop()
cart.sort()
print(cart, len(cart))

# Expected output: ['Latte', 'Water'] 2

#set إزالة الآكرار باسااااام — A2
numbers = [4, 8, 4, 2, 8, 9, 2, 1, 9, 9]
unique = list(set(numbers))
unique.sort()
print(unique)

# Expected output: [1, 2, 4, 8, 9]

#(max / اعال / إآاا فة / أآف :dictionary) ءفار الاراااا — A3
grades = {"math": 88, "english": 76}
grades["math"] = 92
grades["physics"] = 79
del grades["english"]
best = max(grades.values())
print(grades, best)

# Expected output: {'math': 92, 'physics': 79} 92

#tuple (unpacking) فك ّ ال— A4
student = ("نىلى", 21, 92.5)
name, age, score = student
print(name, age, score)
pairs = [("سارة", 85), ("أأما", 77)]
first_name, first_score = pairs[0]
assert name == "نىلى" and age == 21 and score == 92.5
assert first_name == "سارة" and first_score == 85

# Expected output: نىلى 21 92.5

#A5 # إنشاء قاىمة أسعار
prices_a = [10, 20, 30]
prices_b = prices_a.copy()
prices_b.append(40)
print("prices_a:", prices_a)
print("prices_b:", prices_b)
assert prices_a == [10, 20, 30], "prices_a ااعىراا - لازم افضل زى ما هى"
assert prices_b == [10, 20, 30, 40], "prices_b هى اللى المفروض اااا 40"

# Expected output:
# prices_a: [10, 20, 30]
# prices_b: [10, 20, 30, 40]
```
