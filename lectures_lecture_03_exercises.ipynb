```python
grade = 85
level = None

if grade >= 90:
    level = "A"
elif grade >= 80:
    level = "B"
elif grade >= 70:
    level = "C"
elif grade >= 60:
    level = "D"
else:
    level = "F"

print(f"Grade {grade} -> {level}")

assert level in ("A", "B", "C", "D", "F"), "Invalid grade level"
assert not (grade >= 90 and level != "A"), "90-100 should be A"
assert not (80 <= grade < 90 and level != "B"), "80-89 should be B"
assert not (70 <= grade < 80 and level != "C"), "70-79 should be C"
assert not (60 <= grade < 70 and level != "D"), "60-69 should be D"
assert not (grade < 60 and level != "F"), "Below 60 should be F"

print("A1: Passed")
```

```python
import random

target = random.randint(1, 50)
tries = 0

def get_valid_guess(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Please enter a valid integer.")

guess = get_valid_guess("Guess a number from 1 to 50: ")

while guess != target:
    tries += 1

    if guess < target:
        print("Higher")
    else:
        print("Lower")

    guess = get_valid_guess("Try again: ")

tries += 1

print(f"You guessed it! Number of tries: {tries}")

assert tries >= 1, "Don't forget the first attempt"

print("A3: Excellent")
```

```python
scores = [82, 90, 76, 88, 95]

total = 0
count = 0
biggest = scores[0]

for score in scores:
    total += score
    count += 1

    if score > biggest:
        biggest = score

average = total / count

print(total, round(average, 2), biggest)

assert total == 431, "total = sum of the five scores"
assert count == 5, "count should increase in each loop"
assert biggest == 95, "biggest should contain the largest score"
assert abs(average - 86.2) < 0.001, "average = total / count"

print("A4: Excellent")
```
