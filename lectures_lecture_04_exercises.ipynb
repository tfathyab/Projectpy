```python
def is_even(n):
    return n % 2 == 0


def square(n):
    return n * n


def average(values):
    return sum(values) / len(values)


print(is_even(4), square(7))

assert is_even(4) == True and is_even(7) == False, "is_even"
assert square(7) == 49, "square"
assert abs(average([82, 90, 76]) - 82.67) < 0.01, "average"

print("A1: Excellent")
```

```python
def greet(name, greeting="Ahlan"):
    return greeting + ", " + name


print(greet("Sara"))

assert greet("Sara") == "Ahlan, Sara", "default"
assert greet("Omar", "Marhaba") == "Marhaba, Omar", "positional"
assert greet(name="Ali", greeting="Hi") == "Hi, Ali", "keyword"

print("A2: Excellent")
```

```python
def min_max(values):
    return (min(values), max(values))


low, high = min_max([82, 90, 76])
print(low, high)

assert min_max([82, 90, 76]) == (76, 90), "smallest first"
assert min_max([5]) == (5, 5), "single element"
assert min_max([-3, -9, 2]) == (-9, 2), "negative and positive"

print("A3: Excellent")
```

```python
lines = ["date,category", "2026-10-01,food", "2026-10-02,drink"]
path = "my_lines.txt"
content = ""

with open(path, "w") as file:
    for line in lines:
        file.write(line + "\n")

try:
    with open(path, "r") as file:
        content = file.read()
except FileNotFoundError:
    content = "File not found"

print(content)

assert "2026-10-01,food" in content, "first line exists"
assert "2026-10-02,drink" in content, "all lines exist"

print("A4: Excellent")
```
