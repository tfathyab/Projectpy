```python
#(append / insert / remove / pop / sort) ﺳﻠﺔ اﻟﻤﺸﺘﺮﻳﺎت — A1
cart = []
cart.append("Latte")
cart.append("Croissant")
cart.append("Tea")
cart.remove("Croissant")
cart.insert(0, "Water")
cart.pop()
cart.sort()
print(cart, len(cart))
```

```python
#set إزاﻟﺔ اﻟﺘﻜﺮار ﺑﺎﺳﺘﺨﺪام — A2
numbers = [4, 8, 4, 2, 8, 9, 2, 1, 9, 9]
unique = list(set(numbers))
unique.sort()
print(unique)
```

```python
#(max / ﺗﻌﺪﻳﻞ / إﺿﺎﻓﺔ / ﺣﺬف :dictionary) دﻓﺘﺮ اﻟﺪرﺟﺎت — A3
grades = {"math": 88, "english": 76}
grades["math"] = 92
grades["physics"] = 79
del grades["english"]
best = max(grades.values())

print(grades, best)
```

```python
#tuple (unpacking) ﻓﻚ ّ اﻟـ — A4
student = ("نيلي", 21, 92.5)
name, age, score = student

print(name, age, score)
pairs = [("سارة", 85), ("أحمد", 77)]
first_name, first_score = pairs[0]
assert name == "نيلي" and age == 21 and score == 92.5
assert first_name == "سارة" and first_score == 85
```
